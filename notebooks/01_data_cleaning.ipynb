```python
# Import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load raw datasets

customers1 = pd.read_csv('../data/raw/customers.csv')
restaurants1 = pd.read_csv('../data/raw/restaurants.csv')
orders1 = pd.read_csv('../data/raw/orders.csv')
deliveries1 = pd.read_csv('../data/raw/deliveries.csv')
drivers1 = pd.read_csv('../data/raw/drivers.csv')
reviews1 = pd.read_csv('../data/raw/reviews.csv')
payments1 = pd.read_csv('../data/raw/payments.csv')

# Make copies of raw datasets to perform cleaning and feature engineering

customers = customers1.copy()
restaurants = restaurants1.copy()
orders = orders1.copy()
deliveries = deliveries1.copy()
drivers = drivers1.copy()
reviews = reviews1.copy()
payments = payments1.copy()
```

```python
# Store all datasets in a dictionary

datasets = {
    'customers': customers,
    'restaurants': restaurants,
    'orders': orders,
    'deliveries': deliveries,
    'drivers': drivers,
    'reviews': reviews,
    'payments': payments
}
```

```python
# Create a consolidated overview of all datasets

overview = []

for name, df in datasets.items():
    overview.append({
        'dataset': name,
        'rows': df.shape[0],
        'columns': df.shape[1],
        'duplicate_rows': df.duplicated().sum(),
        'missing_values': df.isna().sum().sum()
    })

overview_df = pd.DataFrame(overview)

overview_df
```

```python
# Display columns for every dataset

for name, df in datasets.items():
    print(f'\n{name.upper()}')
    print('-' * 50)
    print(df.columns.tolist())
```

```python
# Display data types for every dataset

for name, df in datasets.items():
    print(f'\n{name.upper()}')
    print('-' * 50)
    print(df.dtypes)
```

```python
# Define primary keys for each dataset

primary_keys = {
    'customers': 'customer_id',
    'restaurants': 'restaurant_id',
    'orders': 'order_id',
    'deliveries': 'delivery_id',
    'drivers': 'driver_id',
    'reviews': 'review_id',
    'payments': 'payment_id'
}

# Check primary-key uniqueness

primary_key_summary = []

for name, primary_key in primary_keys.items():
    df = datasets[name]

    primary_key_summary.append({
        'dataset': name,
        'primary_key': primary_key,
        'unique_values': df[primary_key].nunique(),
        'total_rows': len(df),
        'duplicate_keys': df[primary_key].duplicated().sum(),
        'missing_keys': df[primary_key].isna().sum()
    })

primary_key_df = pd.DataFrame(primary_key_summary)

primary_key_df
```

```python

# Convert date columns to datetime

customers['signup_date'] = pd.to_datetime(
    customers['signup_date'],
    errors='coerce'
)

orders['order_timestamp'] = pd.to_datetime(
    orders['order_timestamp'],
    errors='coerce'
)

# Check for invalid dates after conversion

date_quality = pd.DataFrame({
    'dataset': [
        'customers',
        'orders'
    ],
    'invalid_dates': [
        customers['signup_date'].isna().sum(),
        orders['order_timestamp'].isna().sum()
    ]
})

date_quality
```

```python
# Check whether every order references a valid customer

invalid_customer_ids = orders[
    ~orders['customer_id'].isin(
        customers['customer_id']
    )
]

print(
    'Invalid customer references:',
    len(invalid_customer_ids)
)
```

```python
# Check whether every order references a valid restaurant

invalid_restaurant_ids = orders[
    ~orders['restaurant_id'].isin(
        restaurants['restaurant_id']
    )
]

print(
    'Invalid restaurant references:',
    len(invalid_restaurant_ids)
)
```

```python
# Check whether every order references a valid restaurant

invalid_restaurant_ids = orders[
    ~orders['restaurant_id'].isin(
        restaurants['restaurant_id']
    )
]

print(
    'Invalid restaurant references:',
    len(invalid_restaurant_ids)
)
```

```python
# Check whether every delivery references a valid driver

invalid_driver_ids = deliveries[
    ~deliveries['driver_id'].isin(
        drivers['driver_id']
    )
]

print(
    'Invalid driver references:',
    len(invalid_driver_ids)
)
```

```python
# Identify delivered orders

delivered_orders = set(
    orders.loc[
        orders['order_status'] == 'Delivered',
        'order_id'
    ]
)

delivery_orders = set(
    deliveries['order_id']
)

missing_deliveries = delivered_orders - delivery_orders

print(
    'Delivered orders without delivery records:',
    len(missing_deliveries)
)
```

```python
# Identify cancelled and failed orders

cancelled_failed_orders = set(
    orders.loc[
        orders['order_status'].isin(
            ['Cancelled', 'Failed']
        ),
        'order_id'
    ]
)

invalid_deliveries = (
    cancelled_failed_orders
    & delivery_orders
)

print(
    'Cancelled/failed orders with deliveries:',
    len(invalid_deliveries)
)
```

```python
# Check unique values in categorical columns

categorical_columns = {
    'customers': ['gender', 'city', 'membership_type'],
    'restaurants': ['city', 'cuisine', 'restaurant_category'],
    'drivers': ['gender', 'vehicle_type'],
    'orders': [
        'order_status',
        'payment_method',
        'weather',
        'traffic_condition'
    ],
    'payments': ['payment_method', 'payment_status'],
    'reviews': ['sentiment']
}

for dataset_name, columns in categorical_columns.items():
    print(f'\n{dataset_name.upper()}')
    print('-' * 50)

    df = datasets[dataset_name]

    for column in columns:
        print(f'{column}: {df[column].unique()}')
```

```python
# Check numerical columns for invalid values

print('Customer age outside expected range:')
print(customers[~customers['age'].between(18, 65)])

print('\nRestaurant rating outside expected range:')
print(restaurants[~restaurants['restaurant_rating'].between(1, 5)])

print('\nDriver rating outside expected range:')
print(drivers[~drivers['driver_rating'].between(1, 5)])

print('\nInvalid order amounts:')
print(orders[orders['order_amount'] <= 0])

print('\nInvalid tip amounts:')
print(orders[orders['tip_amount'] < 0])

print('\nInvalid delivery distances:')
print(deliveries[deliveries['delivery_distance_km'] <= 0])
```

```python
# Create final data-quality summary

final_quality_summary = []

for name, df in datasets.items():
    final_quality_summary.append({
        'dataset': name,
        'rows': len(df),
        'columns': len(df.columns),
        'missing_values': df.isna().sum().sum(),
        'duplicate_rows': df.duplicated().sum()
    })

final_quality_df = pd.DataFrame(
    final_quality_summary
)

final_quality_df
```

```python
# Create cleaned copies of all datasets

customers_clean = customers.copy()
restaurants_clean = restaurants.copy()
orders_clean = orders.copy()
deliveries_clean = deliveries.copy()
drivers_clean = drivers.copy()
reviews_clean = reviews.copy()
payments_clean = payments.copy()
```

```python
# Create the processed data directory if it does not exist

from pathlib import Path

processed_dir = Path('../data/processed')

processed_dir.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
# Export cleaned datasets

customers_clean.to_csv(
    processed_dir / 'customers_clean.csv',
    index=False
)

restaurants_clean.to_csv(
    processed_dir / 'restaurants_clean.csv',
    index=False
)

orders_clean.to_csv(
    processed_dir / 'orders_clean.csv',
    index=False
)

deliveries_clean.to_csv(
    processed_dir / 'deliveries_clean.csv',
    index=False
)

drivers_clean.to_csv(
    processed_dir / 'drivers_clean.csv',
    index=False
)

reviews_clean.to_csv(
    processed_dir / 'reviews_clean.csv',
    index=False
)

payments_clean.to_csv(
    processed_dir / 'payments_clean.csv',
    index=False
)
```

```python
# Verify processed datasets

processed_datasets = {
    'customers_clean': customers_clean,
    'restaurants_clean': restaurants_clean,
    'orders_clean': orders_clean,
    'deliveries_clean': deliveries_clean,
    'drivers_clean': drivers_clean,
    'reviews_clean': reviews_clean,
    'payments_clean': payments_clean
}

for name, df in processed_datasets.items():
    print(
        f'{name:<20} {len(df):>6,} rows'
    )
```
